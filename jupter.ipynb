```python
import _mysql_connector
from sqlalchemy import create_engine,text
import pandas as pd
```

```python
engine = create_engine("mysql+mysqlconnector://root:0007@localhost")
```

```python
with engine.connect() as conn:
    conn.execute(text("CREATE DATABASE IF NOT EXISTS Sales_Management_System"))
    conn.commit()
```

```python
db_engine = create_engine("mysql+mysqlconnector://root:0007@localhost/Sales_Management_System")
```

```python
with db_engine.connect() as conn:
    conn.execute(text("""CREATE TABLE IF NOT EXISTS branches (
                      branch_id INT PRIMARY KEY,
                      branch_name VARCHAR(50),
                      branch_admin_name VARCHAR(50)
                      )"""))
    conn.commit()
```

```python
branches = pd.read_csv(r"D:\PROJECTS\Anna_Project_1\Sales_Intelligence_Hub\CSV\branches.csv")
branches.to_sql("branches", db_engine, if_exists="append", index=False)
```

```python
with db_engine.connect() as conn:
    conn.execute(text("""CREATE TABLE IF NOT EXISTS users (
                      user_id INT PRIMARY KEY,
                      username VARCHAR(100),
                      password VARCHAR(255),
                      branch_id INT,
                      role ENUM('Super Admin', 'Admin'),
                      email VARCHAR(255) UNIQUE      
                      )"""))
    conn.commit()
```

```python
users = pd.read_csv(r"D:\PROJECTS\Anna_Project_1\Sales_Intelligence_Hub\CSV\users.csv")
users.to_sql("users", db_engine, if_exists="append", index=False)
```

```python
with db_engine.connect() as conn:
    conn.execute(text("""CREATE TABLE IF NOT EXISTS customer_sales (
                      sale_id INT PRIMARY KEY,
                      branch_id INT ,
                      date DATE ,
                      name VARCHAR(50) ,
                      mobile_number VARCHAR(15),
                      product_name VARCHAR(30),
                      gross_sales DECIMAL(12,2),
                      received_amount DECIMAL(12,2),
                      pending_amount DECIMAL(12,2), 
                      status ENUM('Open','Close')
                      )"""))
    conn.commit()
```

```python
customer_sales = pd.read_csv(r"D:\PROJECTS\Anna_Project_1\Sales_Intelligence_Hub\CSV\customer_sales.csv")
customer_sales.to_sql("customer_sales", db_engine, if_exists="append", index=False)
```

```python
with db_engine.connect() as conn:
    conn.execute(text("""CREATE TABLE IF NOT EXISTS payment_splits  (
                      payment_id INT PRIMARY KEY,
                      sale_id INT ,
                      payment_date DATE ,
                      amount_paid DECIMAL(12,2) ,
                      payment_method VARCHAR(50)
                    )"""))
    conn.commit()
```

```python
payment = pd.read_csv(r"D:\PROJECTS\Anna_Project_1\Sales_Intelligence_Hub\CSV\payment_splits.csv")
payment.to_sql("payment_splits", db_engine, if_exists="append", index=False)
```
